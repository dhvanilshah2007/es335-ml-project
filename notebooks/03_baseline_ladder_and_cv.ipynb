### Key Questions Investigated in Phase 4
1. **Progression Across the Baseline Ladder (Rungs 0 to 5):** How much does predictive performance improve as we climb from the Mean/Prior floor (Rung 0), Fixed Schedule $K^*$ (Rung 1), Age-Only (Rung 2), Instantaneous Sensors (Rung 3), Trailing Window Linear Models (Rung 4), up to Non-Linear Tree Ensembles (Rung 5)?
2. **NASA Asymmetric Penalty & Near-Failure Precision:** Which models minimize dangerous late predictions (overestimating RUL when the engine is close to failure)?
3. **Probability Calibration & Threshold Optimization:** Are predicted alarm probabilities $P(\text{RUL} \le 30)$ well-calibrated, and what optimal threshold $p^*_{\text{dev}}$ minimizes expected maintenance cost?
4. **Feature Attribution & Sparsity:** Which trailing window features drive `HistGradientBoosting` vs. $L_1$-penalized `Lasso`?
